```python
def solveHanoi(numOfdisks):
    hanoi = [[], [], []]
    for x in range(numOfdisks, 0, -1):
        hanoi[0].append(x)

    print(hanoi)

    while len(hanoi[-1])!=numOfdisks:
        if len(hanoi[1])==0:
            hanoi[1].append(hanoi[0].pop())
        if  len(hanoi[-1])==0 or hanoi[-1][-1] > hanoi[1][-1]:
            hanoi[-1].extend(hanoi[1])
            hanoi[1].clear()
        else:
            hanoi[1].extend(hanoi[-1])
            hanoi[-1].clear()


    print(hanoi)
```

```python
solveHanoi(10)
```

```python
solveHanoi(3)
```

```python
def solveHanoi0(numOfdisks, numOfSticks):
    hanoi = []
    for i in range(numOfSticks):
        hanoi.append([])
    for x in range(numOfdisks, 0, -1):
        hanoi[0].append(x)

    print(hanoi)

    while len(hanoi[-1])!=numOfdisks:
        if len(hanoi[1])==0:
            hanoi[1].append(hanoi[0].pop())
        for j in range(1, len(hanoi)-1):
            if  len(hanoi[j+1])==0 or hanoi[j+1][-1] > hanoi[j][-1]:
                hanoi[j+1].extend(hanoi[j])
                hanoi[j].clear()
            else:
                hanoi[j].extend(hanoi[j+1])
                hanoi[j+1].clear()


    print(hanoi)
```

```python
solveHanoi0(10, 5)
```

```python
#with the start and end points defined

def solveHanoi1(numOfdisks, numOfSticks, start, end):
     hanoi = []
     for i in range(numOfSticks):
         hanoi.append([])
     for x in range(numOfdisks, 0, -1):
         hanoi[start-1].append(x)

     print(hanoi)

     first = hanoi[start-1]
     hanoi[start-1] = []
     hanoi[end-1] = first
    
     print(hanoi)
```

```python
solveHanoi1(10, 5, 2, 1)
```
